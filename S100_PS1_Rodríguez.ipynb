> pares = pares[pares < 0.999]                                       # fuera la diagonal

````python
# a) Matriz de correlación
corr = ...
# a) Matriz de correlación de las 10 variables de audio

corr = df[audio].corr()

print("Matriz de correlación:")
display(corr.round(3))
````

````python
# b) Lista de pares ordenada: el más positivo y el más negativo

pares = corr.unstack()

# Eliminar la diagonal
pares = pares[pares < 0.999]

# Eliminar pares repetidos de la matriz simétrica
pares = pares[
    pares.index.get_level_values(0)
    < pares.index.get_level_values(1)
]

# Ordenar desde la correlación más negativa hasta la más positiva
pares = pares.sort_values()

print("Cantidad de pares:", len(pares))
print("\nLista ordenada de pares:")
display(pares.to_frame(name="correlacion").round(3))

par_mas_negativo = pares.index[0]
r_mas_negativo = pares.iloc[0]

par_mas_positivo = pares.index[-1]
r_mas_positivo = pares.iloc[-1]

print("\nPar más negativo:", par_mas_negativo)
print("Correlación:", round(r_mas_negativo, 3))

print("\nPar más positivo:", par_mas_positivo)
print("Correlación:", round(r_mas_positivo, 3))

````

````python
# c) Diagrama de dispersión del par más positivo
````

````python
# c) Diagrama de dispersión del par más positivo


variable_x, variable_y = par_mas_positivo

# Muestra reproducible de 4,000 canciones
m = df.sample(4000, random_state=1)

plt.figure(figsize=(8, 5))

plt.scatter(
    m[variable_x],
    m[variable_y],
    s=6,
    alpha=0.25,
    color=AZUL,
    linewidths=0
)

plt.xlabel(variable_x)
plt.ylabel(variable_y)
plt.title(
    f"Relación entre {variable_x} y {variable_y}\n"
    f"Correlación de Pearson: r = {r_mas_positivo:.3f}"
)

plt.show()

````

````python
# d) Correlación de cada variable de audio con track_popularity

corr_popularidad = (
    df[audio + ["track_popularity"]]
      .corr()["track_popularity"]
      .drop("track_popularity")
      .sort_values()
)

print("Correlaciones con la popularidad:")
display(corr_popularidad.to_frame(name="correlacion").round(3))

# Variable con la relación más fuerte, independientemente del signo
variable_mas_relacionada = corr_popularidad.abs().idxmax()
valor_correlacion = corr_popularidad.loc[variable_mas_relacionada]

print(
    "\nVariable más relacionada con track_popularity:",
    variable_mas_relacionada
)
print("Correlación:", round(valor_correlacion, 3))
````

````python
# b) La misma correlación dentro de cada género
#corr_genero = ...
#print(corr_genero)


print("Código de b")

corr_genero = (
    df.groupby("playlist_genre")
      .apply(lambda g: g["energy"].corr(g["danceability"]))
      .sort_values()
)

print("\nCorrelación por género:")
print(corr_genero)
````

````python
# c) Tabla resumen por género

print("\nCódigo de c")

tabla_resumen = (
    df.groupby("playlist_genre")["energy"]
      .agg(["count", "mean", "median", "std", "min", "max"])
)

# Agregar la correlación energy-danceability
tabla_resumen["correlacion"] = corr_genero

print("\nTabla resumen por género:")
print(tabla_resumen)

````

````python
# d) Estandarización de energy y canción más extrema

print("\nCódigo de d")

# Calcular z-score
df["energy_z"] = (
    (df["energy"] - df["energy"].mean())
    / df["energy"].std()
)

# Canción con el z-score más alto
i = df["energy_z"].idxmax()

print("\nCanción con energy más extrema (z más alto):")
print(
    df.loc[i, [
        "track_name",
        "track_artist",
        "energy",
        "energy_z"
    ]]
)

# Comprobaciones
print("\nMedia de energy_z:", df["energy_z"].mean())
print("Desviación de energy_z:", df["energy_z"].std())
print("Skew original:", df["energy"].skew())
print("Skew estandarizado:", df["energy_z"].skew())
````
